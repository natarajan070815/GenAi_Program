#Loops


```python
#while loop(also called as Entry control loop)- initialization;condition;increment/Decrement
#print value from 1 to 20
i=1
while(i<=20):

  print(i)
  i=i+1


```

```python
#print value 20 to 1
a=20
while(a>=1):
  print(a)
  a-=1
```

#For Loop


#For Loop with list

```python
fruits=["apple","banana","cherry"]
for i in fruits:
  print(i)
```

#For loop with Range


```python
#for loop with range(start,stop-1,step)
#print values from 1 to 10
for value in range (1, 11, 1):
  print(value)
```

```python
# print all the even number between 1 to 20 using range function
for even in range (2,21,2):
  print(even)
```

```python
#print 1 to 10 (by default step value consider as 1)
for step in range (1,11):
 print(step)
```

```python
#print 0 to 5
for value1 in range (6):
  print(value1)
```

```python
#print number from 22 to 11 in reverse order
for number in range(22,12,-1):
  print(number)
```

#For loop with Conditional Statement

```python
# values from 1 to 10 but it should be display as odd or even
for value in range(1, 11):
    if value % 2 == 0:
        print(f"{value} is even")
    else:
        print(f"{value} is odd")
```

```python
# print from 1 to 10 but dont need to 5
for values in range (1,11,1):
  if(values!=5):
      print(values)

```

```python
# print from 1 to 10 but dont need to 5 #pass means do nothing
for values in range (1,11,1):
  if(values==5):
      pass #pass means do nothing
  else:
    print(values)
```

```python
# print from 1 to 10 but dont need to 5
for values in range (1,11,1):
  if(values==5):
      continue #contine means it will skip
  else:
    print(values)
```

```python
# print 5,15,20,25,30
for i in range(5,31,5):
  if(i==10):
    continue
    print(i)
```

```python
# print from 1 to 10
for values in range (1,11,1):
  if(values==5):
      break #break means it will stop
  else:
    print(values)
```

```python
# print 1 to 100
for i in range (1,101,1):
  if(i==21):
    break
  print(i)
```
