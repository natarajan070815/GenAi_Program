```python
# Create a dictionary with key-value pairs
student = {"name": "Nataraj", "city": "Chennai"}
print(student)
```

```python
# Write a Python program to find the largest number in a list
numbers = [12, 45, 7, 89, 34, 2, 45, 12, 60, 7]
largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print("Largest number:", largest)
```

```python
# Write a Python program to count the frequency of elements in a list
numbers = [12, 45, 7, 89, 34, 2, 45, 12, 60, 7]
freq = {}
for num in numbers:
    freq[num] = freq.get(num, 0) + 1
print("Frequency:", freq)
```

```python
# Create a dictionary that stores student names as keys and marks as values, then find the student with the highest marks
marks = {
    "Asha": 88,
    "Ravi": 92,
    "Meena": 79,
    "Kumar": 85,
    "Divya": 67,
    "Priya": 45
}

top_student = None
highest = -1
for name, mark in marks.items():
    if mark > highest:
        highest = mark
        top_student = name
print("Topper:", top_student, "with", highest, "marks")
```

```python
# Implement a while loop to print the Fibonacci series up to a given number
limit = int(input("Enter the limit for Fibonacci: "))
a, b = 0, 1
print("   Fibonacci series:", end=" ")
while a <= limit:
    print(a, end=" ")
    a, b = b, a + b
print()
```

```python
# Use a set to remove duplicate elements from a list
numbers = [12,12,45,7,7,89,34,2,2,45,12,60,7]
unique = list(set(numbers))
print("Without duplicates:", unique)
```
