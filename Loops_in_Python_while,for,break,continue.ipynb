# While Loop

```python
i=1
while i<=20:
    print(i)
    i=i+1
```

```python
# print values from 20 to 1
i1=20
while(i1>=1):
  print(i1)
  i1-=1
```

# For Loop with Range

```python
for i in range(1,11,1):
  print(i)
```

```python
# print all the even number between 1 to 20
for i in range(1,21,1):
  if i%2==0:
    print(i)

```

```python
# print from 1 to 10
for i in range(11):
  print(i)
```

```python
#print values from 0 to 5
for i in range(6):
  print(i)
```

```python
# print numbers from 22 to 11
for i in range(22,10,-1):
  print(i)
```

# For Loop with conditional Statement

```python
# print values from 1 to 10 it should be display as odd or even
for i in range(1,11):
  if i%2==0:
    print(f"{i} is a even number")
  else:
    print(f"{i} is an odd number")
```

```python
# print 1 to 10 but dont want print 5
for i in range(1,11):
  if(i!=5):
     print(i)
```

# Pass Keyword

```python
for i in range(1,11):
  if(i==5):
    pass # do nothing
  else:
    print(i)
```

# Continue Keyword

```python
for i in range(1,11):
  if(i==5):
    continue # skip the current iteration
  else:
    print(i)
```

```python
# print 5,15,20,25,30
for i in range(5,31,5):
  if i==10:
    continue
  else:
   print(i)
```

# Break Keyword

```python
for i in range(1,11):
  if(i==5):
    break # stop the current iteration
  else:
    print(i)
```

```python
# Print Values from 1 to 100 but it should stop at value 20
for i in range(1,101):
  if(i==21):
    break
  else:
     print(i)
```
