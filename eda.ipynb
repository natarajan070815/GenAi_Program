# Outliers Detection in total bill column


```python
Q1=df['total_bill'].quantile(0.25)
Q3=df['total_bill'].quantile(0.75)
print(Q1)
print(Q3)
```

```python
IQR=Q3-Q1 #IQR means Inter Quantile Range
```

```python
IQR
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# 1. Calculate Outliers using IQR
Q1 = df['total_bill'].quantile(0.25)
Q3 = df['total_bill'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_lower = df[df['total_bill'] < lower_bound]['total_bill'].tolist()
outliers_upper = df[df['total_bill'] > upper_bound]['total_bill'].tolist()

print(f"Q1 (25th percentile): {Q1:.4f}")
print(f"Q3 (75th percentile): {Q3:.4f}")
print(f"Lower Bound for Outliers (Q1 - 1.5*IQR): {lower_bound:.4f}")
print(f"Upper Bound for Outliers (Q3 + 1.5*IQR): {upper_bound:.4f}")
print(f"Outliers below lower bound: {outliers_lower}")
print(f"Outliers above upper bound (High Outliers): {outliers_upper}")

# 2. Draw a distribution plot (with shaded area under the curve) representing total_bill
plt.figure(figsize=(10, 6))
sns.histplot(df['total_bill'], kde=True, stat="density", color="skyblue", fill=True)
plt.title('Distribution of Total Bill with Density Curve')
plt.xlabel('Total Bill')
plt.ylabel('Density')
plt.axvline(Q1, color='red', linestyle='--', label=f'Q1: {Q1:.2f}')
plt.axvline(Q3, color='orange', linestyle='--', label=f'Q3: {Q3:.2f}')
plt.axvline(upper_bound, color='purple', linestyle=':', label=f'Upper Limit: {upper_bound:.2f}')
plt.legend()
plt.show()
```
