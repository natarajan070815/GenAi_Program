# if elif else Ladder


```python
# Creating Grading System
# 90 >= 'A' , 75 to 90 'B' , 60 to  75 'C' , 35 to 60 'D' , 20 to 35 'E' , 0 to 20 'F'

mark = 85
if (mark >=90) and (mark <=100):
  print("A")
elif (mark >=75) and (mark <90):
  print("B")
elif (mark >=60) and (mark <75):
  print("C")
elif(mark >=35) and (mark <60):
  print("D")
elif (mark >=20) and (mark <35):
  print("E")
elif (mark >=0) and (mark <20):
  print("F")
else:
  print("Invalid Input")
```

# Nested Condition

```python
# Voting
age=25
indian_citizen= True

if (age >= 18 and indian_citizen):
  print("Person is indian citizen and >=18 years. So eligible for vote")
else:
  print("You are not eligible for vote")
```
