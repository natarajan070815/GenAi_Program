# Tuples
 # Creating tuples


```python
tuple1=("apple","banana","Cherry")
tuple2=(1,2,3,4)
tuple3=("text",42,True)
```

```python
print(tuple1)
print(tuple2)
print(tuple3)
```

# Accessing the element in the tuple

```python
 # Accessing apple
tuple1[0]
```

```python
 # Accessing cherry
tuple1[-1]
```

```python
 # Accessing text,42
tuple3[0:2] # using range
```

```python
# Accessing 2,3,4
tuple2[1:] # using range but not required to specify the end value because it is last value
```

# Tuple Operations

```python
# length
len(tuple2)
```

```python
# type
type(tuple2)
```

```python
# Concatenation
tuple1 + tuple2
```

```python
# repeatation
tuple2*2
```

```python
tuple1*3
```

# Dictionary


# Create a dictionary with keys name,age,job and the values alice,25,engineer

```python
dict1={"name":"alice","age":25,"job":"engineer"}
```

```python
dict1
```

```python
empty=dict()
```

```python
empty
```

```python
type(empty)
```

# Accessing Values from the dictionary

```python
dict1["name"]
```

```python
dict1['job']
```

# Updating the values in the dictionary

```python
dict1
```

```python
# updating age from 25 to 26
dict1["age"]=26
```

```python
dict1
```

# Adding the key value pair

```python
# Add key value pair as country:india
dict1["country"]="india"
```

```python
dict1
```

# Removing the key value pair

```python
dict1
```

```python
# removing the key value pair job:engineer
del dict1["job"]
```

```python
dict1
```

# Common Functions in Dictionary

```python
dict1.keys()
```

```python
dict1.values()
```

```python
dict1.items() # it will displays keys and values available in the dictionary
```

# Looping through dictionary

```python
for i in dict1.items(): # print the values with key value pair in for loop
    print(i)
```
