#importing Necessary Lbaries

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

# Loading the Dataset

```python
df=pd.read_csv('/content/tips.csv')
```

```python
df
```

# Data Visualization operations

```python
# 1. Count of Record by day
df['day'].value_counts()
```

```python
87+76+62+19
```

```python
df.shape
```

```python
sns.countplot(x='day',data=df,color='red')
plt.title('Count of Record by day')
plt.show()
```

# Bar Chart used for categorical data

 Inference : Saturday has more number of Customers

# Total Bill Trendover Time


```python
df.head()
```

```python
df['total_bill'].plot()
plt.title("Total Bill Trendover Time")
plt.xlabel("Index")
plt.ylabel("Total Bill")
plt.show()
```

Inference: Peat at 50, Lowest below 10 and Average is 15- 25

# Distribution of TotalBill


```python
sns.histplot(df['total_bill'],kde=True)
plt.title("Distribution of TotalBill")
plt.show()
```

Inference: Most of the people total bill amount lies between  10 -30

Total bil by day

```python
df[df['day']=='Sun']['total_bill'].sum()
```

```python
df.groupby('day')['total_bill'].sum()
```

```python
sns.boxplot(x='day',y='total_bill',data=df)
plt.title("Total Bill by day")
plt.show()
```

Inference: Everyday Average Total Bill lies between 15 - 25

find the relation between total bill and tip

```python
sns.scatterplot(x='total_bill',y='tip',data=df)
plt.title("Total Bill Vs Tip")
plt.show()
```

Inference: If Total bill increased tip value will also increase(positive Correlation)

Realtionship between all the numerical column

```python
df
```

```python
df_corr=df.corr(numeric_only=True)
```

```python
df_corr
```

```python
sns.heatmap(df_corr,annot=True)
plt.title("Correlation Matrix")
plt.show()
```

Inference:  Total bill and tip has the most number of correlation
