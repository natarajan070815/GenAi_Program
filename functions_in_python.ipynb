# Playing with parameters


```python
# addition of two numbers
def addition_value(a1,a2):
  print("Enter Value1 :",a1)
  print("Enter Value2 :",a2)
  a3 = a1 + a2
  print("Addition of two values are :",a3 )
```

```python
addition_value(10,20) # positional arguments
```

```python
addition_value(20,10)
```

```python
addition_value(a2=20,a1=10) #keyword argument
```

```python
def addition_value(a1,a2=10): # default parameter
  print("Enter Value1 :",a1)
  print("Enter Value2 :",a2)
  a3 = a1 + a2
  print("Addition of two values are :",a3 )
```

```python
addition_value(15)
```

```python
addition_value(20,40) #overrides default parameter value
```

```python
def addition_value(a1=15,a2): # default parameter
  print("Enter Value1 :",a1)
  print("Enter Value2 :",a2)
  a3 = a1 + a2
  print("Addition of two values are :",a3 )
```

# Return Keyword

```python
from binascii import b2a_uu
# multiplication of two numbers
# return keyword
def multiplication_value(b1,b2):
  print("Enter Value1 :",b1)
  print("Enter Value2 :",b2)
  b3 = b1 * b2
  print("Multiplication of two values are :",b3 )
  return b3
```

```python
m=multiplication_value(5,5)
```

# Assessment1
Write a function that returns the total cost of the purchase. parameters required are price and quantity


```python
def purchase(price,quantity):
  print("Enter Price :",price)
  print("Enter Quantity :",quantity)
  cost= price * quantity
  print("Total cost for the purchase is :" , cost)
  return cost
```

```python
purchase(10,2)
```
